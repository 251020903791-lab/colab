```python
GPA = float(input("Nhập điểm GPA:"))
if GPA < 0 or GPA >10 :
  print("Vui lòng nhập lại GPA")
else :
  if GPA >= 8.5 :
    print("Học sinh xuất sắc")
  elif 7.0 <= GPA < 8.5 :
      print("Học sinh giỏi")
  elif 5.5 <= GPA < 7.0 :
      print("Học sinh khá")
  elif 4.0 <= GPA < 5.5 :
      print("Học sinh trung bình")
  else:
    print("Học sinh yếu")
```

```python

n = int(input("Nhập một số nguyên dương n: "))


tong = 0


for i in range(1, n + 1):
    if i % 2 == 0:
        tong += i


print("Tổng các số chẵn từ 1 đến", n, "là:", tong)
```

```python
import getpass
mk_dung ="chao ban hoc"
so_lan = 0
max_lan = 3
while so_lan < max_lan :
  mk = getpass.getpass("Nhập mật khẩu:")
  so_lan += 1

  if mk == mk_dung :
    print("Đăng nhập thành công")
    break
  else:
    if so_lan < max_lan :
      print("Sai mật khẩu")
    else:
      print("Khóa tài khoản")
```
