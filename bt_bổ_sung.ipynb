```python
dem_so_duong = 0
dem_so_am = 0
dem_so_khong = 0

# Số lượng số cần nhập
so_luong = 10

print(f"Vui lòng nhập {so_luong} số nguyên:")

# Bắt đầu vòng lặp để nhập 10 số
for i in range(1, so_luong + 1):
    while True:
        try:
            # Nhập số thứ i
            so_nhap = input(f"Nhập số thứ {i}: ")
            so = int(so_nhap)
            break  # Thoát khỏi vòng lặp while nếu nhập thành công
        except ValueError:
            print("Lỗi: Vui lòng chỉ nhập SỐ NGUYÊN.")

    # Tiến hành kiểm tra và đếm
    if so > 0:
        dem_so_duong += 1  # Tăng biến đếm số dương
    elif so < 0:
        dem_so_am += 1     # Tăng biến đếm số âm
    else: # Trường hợp còn lại (so == 0)
        dem_so_khong += 1  # Tăng biến đếm số không

# In ra kết quả cuối cùng
print("\n--- KẾT QUẢ ĐẾM ---")
print(f"Tổng số đã nhập: {so_luong}")
print(f"Số lượng số dương (> 0): {dem_so_duong}")
print(f"Số lượng số âm (< 0): {dem_so_am}")
print(f"Số lượng số bằng 0 (= 0): {dem_so_khong}")
print("--------------------")
```

```python
import math

# Yêu cầu người dùng nhập số nguyên n
n_str = input("Nhập vào một số nguyên n: ")

try:
    n = int(n_str)

    # 1. Xử lý các trường hợp đặc biệt (n < 2)
    if n < 2:
        print(f"Số {n} KHÔNG phải là số nguyên tố.")

    # 2. Xử lý trường hợp n = 2 (Số nguyên tố chẵn duy nhất)
    elif n == 2:
        print(f"Số {n} là SỐ NGUYÊN TỐ.")

    # 3. Xử lý các trường hợp n > 2
    else:
        # Nếu n là số chẵn lớn hơn 2, chắc chắn không phải số nguyên tố
        if n % 2 == 0:
             print(f"Số {n} KHÔNG phải là số nguyên tố (vì chia hết cho 2).")
        else:
            # Thuật toán kiểm tra số nguyên tố:
            # Ta chỉ cần kiểm tra các ước số từ 3 đến căn bậc hai của n
            # (chỉ kiểm tra các số lẻ)

            # Tính căn bậc hai của n (làm tròn xuống)
            can_bac_hai = int(math.sqrt(n))

            la_so_nguyen_to = True # Giả định ban đầu là số nguyên tố

            # Bắt đầu kiểm tra từ số 3, với bước nhảy là 2 (chỉ kiểm tra số lẻ)
            for i in range(3, can_bac_hai + 1, 2):
                if n % i == 0:
                    la_so_nguyen_to = False # Tìm thấy ước số, không phải số nguyên tố
                    break # Thoát vòng lặp ngay lập tức

            if la_so_nguyen_to:
                print(f"Số {n} là SỐ NGUYÊN TỐ.")
            else:
                print(f"Số {n} KHÔNG phải là số nguyên tố.")

except ValueError:
    print("Lỗi: Đầu vào không hợp lệ. Vui lòng chỉ nhập số nguyên.")
```
