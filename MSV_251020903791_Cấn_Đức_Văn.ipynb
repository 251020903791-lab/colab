```
n = int(input("Nhập số nguyên dương:"))
tong = 0
for i in range(1, n+1):
  if i % 2 == 0 :
    tong += i
print("Tổng cần tìm là:",tong)
```

```
n = str(input("Nhập chuỗi dữ liệu:"))

words = n.split()
so_tu = len(words)
if words :
  tu_dai_nhat = max(words, key = len)
  print(f"words:{so_tu}", f"Longest words:{tu_dai_nhat}")
else:
 print("Chuỗi rỗng")
```

```
input_numbers_str = input("Nhập danh sách các số nguyên:")
so_nguyen = [int(x) for x in input_numbers_str.split()]
so_la = list(set(so_nguyen))
so_la.sort()

if len(so_la) >= 2 :
    second_largest = so_la[-2]
    print(f"Danh sách các số nguyên duy nhất và đã sắp xếp: {so_la}")
    print(f"Số lớn thứ hai: {second_largest}")
elif len(so_la) == 1:
    print(f"Danh sách các số nguyên duy nhất và đã sắp xếp: {so_la}")
    print(f"Chỉ có một số nguyên duy nhất: {so_la[0]}. Không có số lớn thứ hai.")
else :
    print("Danh sách rỗng. Không có số nguyên nào.")
```

```
input_scort = input("Nhập danh sách điểm :")
scort = [float(x) for x in input_scort.split()]

diem_so = [s for s in scort if 0 <= s <= 10]
if diem_so :
    diem_trung_binh = sum(diem_so) / len(diem_so)
    hoc_sinh_do = [s for s in diem_so if s >= 5]
    so_hoc_sinh_do = len(hoc_sinh_do)

    print(f"Điểm trung bình: {diem_trung_binh:.1f}")
    print(f"Số học sinh qua môn: {so_hoc_sinh_do}")
else:
  print("Không có điểm hợp lệ (0-10) để tính toán.")
```
